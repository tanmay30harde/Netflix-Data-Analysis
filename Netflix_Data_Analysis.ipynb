## **Finding Missing Values**

```python
df.isnull().sum()
```

```python
# Fill missing categorical values
columns_to_fill = [
    'director',
    'cast',
    'country',
    'rating',
    'duration'
]

for col in columns_to_fill:
  df[col] = df[col].fillna("NA")

# Check missing values again
print(df.isnull().sum())
```

## **Fix mixed-type columns (e.g., duration stored as "90 min")**


```python
df['duration_value'] = pd.to_numeric(
    df['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)
df['duration_unit'] = df['duration'].str.extract(
    r'([A-Za-z]+)'
)
df[['title', 'duration', 'duration_value', 'duration_unit']].head(10)
```

## **Parse date columns into proper datetime objects**

```python
df['date_added'] = pd.to_datetime(
    df['date_added'],
    format='%B %d, %Y',
    errors='coerce'
)
print(df['date_added'].dtype)

df[['title', 'date_added']].head()
```

## **Export the cleaned DataFrame to a new CSV file**

```python
df.to_csv(
    '/content/netflix_titles_cleaned.csv',
    index=False
)
```
